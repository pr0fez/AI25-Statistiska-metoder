### Statistik, maskininlärning och regression

I _statistik maskininlärning_ kräver användningen fortfarande starka statistiska konfidensintervall för att vara anvädningsbara. Till exempel inom fysikalisk simulering-- resultatet måste i slutändan vara konsistent med tydliga intervall och övre gränser. Här används regressioner för att hitta _best-fit_ bland en uppsättning fysiska modeller gentemot datan. Även när neurala nätverk används måste resultatet fortfarande analyseras med statistiska metoder för att vara användbart. 

Ett exempel är en nyligen färdigställd simulation av en hel galax från där 100 miljarder individuella stjärnor simulerades över en miljon år på bara 3 timmars beräkning på en superdator. Problemet har länge varit att supernovor är extremt korta händelser i perspektivet över miljontals eller miljarder år, men har en enorm påverkan på hela galaxen. 

Numeriska simulationer skulle behöva gå ner på en upplösning på dagar varje gång en supernova skedde-- helt orealistiskt om en hel galax skall simuleras med hundratals miljarder stjärnor över miljarder år. Så ett neuralt nätverk tränades på att simulera supernovor och tog hand om de händelserna i simulationen. Men varje supernova behövde då (automatiskt) analyseras med avseende på energi- och momentbevarande. Denna hyfsat komplexa statistiska analys var dock _mycket_ lättare att beräkna!

I många andra maskininlärningsproblem, särskilt de som nuvarande entusiasm handlar om, så har vi inte tillgång till sådan statistik och ingen simluation att jämföra resultatet mot. Syftet är istället att reagera på helt okänd data och att producera användbara resultat även när inget omedelbart korrekt svar finns. Men vi behöver förstås fortfarande något sätt att avgöra hur bra det går, annars har vi inget att matematiskt optimera mot!

### Träning, Validering och Testning

I generell maskininlärning tittar vi istället på _trender_. För att spåra detta behöver vi dela upp vår data i flera delmängder. Det första är den _kända datan_, som vi använder för att hitta våra skattingar. Vi _tränar_ modellen på denna data, som oftast benämns just _träningsdata_. 

Under träningen behöver vi även utvärdera utfallet av träningen, för att se trenden. Detta gör vi genom att _validera_ träningen med en andra delmängd, _valideringsdata_. I nästa kurs kommer vi gå igenom flera tekniker för att maximera användingen av data, till exempel _korsvalidering_ som tränar på kombinatoriska urval av validerings- och träningsdata. 

Vi följer alltså trenden mot bättre resultat enligt något kriterium vi satt upp, ett optimeringsvillkor som i maskininlärning kallas en _kostnadsfunktion_ (eng. _loss-function_, _cost-function_, _criterion_). 

I allmänhet formuleras alltid ML problem som ett _minimerings_ problem, men undantag finns. Det går alltid att formulera om ett maximerings- eller jämviktsvillkor till ett minimeringsvillkor, även om det ibland blir mer svårbegripigt. Det är heller inte nödvändigtvis lätt (beräkningsmässigt) att formulera om även om det är möjligt, därav undantagen.

Eftersom vi inte kan lita på antagandena om normalfördelning i allmän maskininlärning måste vi i slutändan utvärdera det slutliga resultatet på helt okänd data, som inte förekommit i något tidigare steg. Detta är den tredje delmängden, _testdatan_. 

I allmänhet spåras ett antal storheter i ML, som har statistisk betydelse:
$$\mathrm{MAE} = \frac{1}{m}\sum_{i=1}^m|{y_i - \hat{y}_i}|$$
$$\mathrm{MSE} = \frac{1}{m}\sum_{i=1}^m({y_i - \hat{y}_i})^2$$
$$\mathrm{RMSE} = \sqrt{\frac{1}{m}\sum_{i=1}^m({y_i - \hat{y}_i})^2}$$

Notera att inga av dessa tillämpar statistika korrektioner och därmed är vare sig RMSE eller MAE rättvisa skattningar. För statistiska regressioner används korrektioner och även _förklaringsgrad_ spelar stor roll för att förfina modellen. Mer om detta nästa gång. 

MSE _är_ rättvis, men som vi sett säger själva storheten inte så mycket. Men för att spåra en trend duger den utmärkt! Det är också ett enkelt optimeringsvillkor som har bra egenskaper: en summa av kvadrater är _kontinuerlig_, _positiv_ och _differentierbar_. Alltså är 0 ett uppenbart optima. För minstakvadrat metoden är det tydligt vad det innebär: om vi når 0 har vi hittat en exakt lösning.

OBS: även om det ser ut som att MSE kan gå mot noll när stickprovet blir stort nog, så är det bara sant om datan normalfördelar kring regressionslinjen-- det vill säga om våran modell faktiskt motsvarar grundsanningen. I allmän ML kan vi inte göra det antagandet, vi kan inte förlita oss på centrala gränsvärdessatsen annat än som en försäkran om att vi hittar _något_ optima. 

MAE och MSE är exempel på _normer_ från linjär algebra. MAE är $\mathcal{l}_1$ normen och MSE är $\mathcal{l}_2$. Mer om detta nästa kurs, vi kommer använda detta för att förbättra våra resultat avsevärt! 

Om datan utgör ett vektorrum så är RMSE en förfördelad skattare (_biased estimate_) för det euklidiska medelavståndet från förutsägelserna till träningsdatan-- även för okänd data. Biaset är felet i träningsdatan och förklarar hur vi kan skatta även okänd data: vi har uttryckligen inkluderat ett okänt bias bort från grundsanningen, och den okända datan följer grundsanningen. Så lite av en icke-lösning, men det tillåter faktiskt att vi går vidare med analysen.

### Förväntat fel på okänd data

I maskininlärning är fördelningen för _felet_, dvs $y_i - \hat{y_i}$, av särksild betydelse. Dess väntevärden tillåter att vi gör förutsägelser _utanför_ träningsdatan. Särskilt gäller det att _variansen_ i felet kan brytas ner i termer av MSE på _okänd data_:

$$\mathbb{E}[(y - \hat{y})^2] = \mathrm{Bias}^2 + \mathrm{Var} + \sigma^2$$

Väntevärdet är ju medlet, så variansen för felet är just MSE. Bias är väntevärdet för regressionslinjen jämfört med grundsanningen-- som är okänd-- och variansen är för datans spridning runt regressionslinjen. Den sista variansen är den _sanna variansen_, även kallat det _irreducibla felet_ som alltså beror på verklighetens brusiga natur och inte något mätfel eller bristfällig modell. Den är förstås också okänd!

Detta har en något olycklig konsekvens: Vi kan egentligen bara påverka variansen, men bias och brus har en mycket större påverkan på resultatet. Dessutom är allt utom den sista termen ett resultat av vår modell, så om vi minskar varians ökar bias! Detta kallas _bias-variance tradeoff_ och sätter hårda gränser på hur bra regressionsmodeller kan bli när det gäller okänd data.

NOTERA:

Low accuracy = High bias

Low precision = High variance

High accuracy = Low bias

High precision = Low variance

Det kan löna sig att tillåta högre varians (lägre precision) för att få lägre bias (högre noggrannhet).

Det kan också löna sig att slå ihop flera regressioner till en stor modell och dra nytta av den större datamängden-- en till, något förvånande konsekvens av bias-varians förhållandet. En stor modell behöver bara göra kompromissen en gång, medans vardera enskild modell, löst uttryckt, har hela felet i vardera modell.
